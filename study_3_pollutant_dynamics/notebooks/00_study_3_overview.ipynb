# Study 3: Pollutant Dynamics

Study 3 is a synthetic annual environment for four U.S. Census regions. It develops pollutant-specific dynamics while keeping Study 2 unchanged as the baseline. The four regions are state groupings, not shared airsheds or watersheds, so each pollutant uses its own spatial assumptions.

Notebooks 05?09 now cover pollutant dynamics, regional economic differences, synthetic income damage, delayed mitigation, and an integrated fixed-policy environment. Stage 5 audits the integrated transitions and baseline policies. No Study 3 agents have been trained, and the reward function has not been selected.

## What needs to be defined

The four pollutants use different measurements and behave differently. PM2.5 is measured as annual air concentration. CO2 is shared across the atmosphere. Nitrate-N moves through water systems. Lead persists in soil, and cleanup can remove it or make it less mobile.

Before using a number, I need to record its source, unit, pollutant form, medium, time period, and any conversion. If a published value does not fit this model directly, I will label it as an assumption.

## Sources from the initial review

These sources describe pollutant processes and measurements. They do not provide a complete parameter set for this model.

- [WHO air quality guidelines](https://www.who.int/publications/b/59593) give health-based PM2.5 guidance.
- [EPA AP-42](https://www.epa.gov/air-emissions-factors-and-quantification/ap-42-compilation-air-emissions-factors-stationary-sources) describes emissions factors for stationary sources.
- [EPA greenhouse gas emission factors](https://www.epa.gov/climateleadership/ghg-emission-factors-hub) supports activity-based greenhouse gas estimates.
- [IPCC carbon-cycle FAQ](https://www.ipcc.ch/site/assets/uploads/2020/05/WGI_AR5_FAQ_EN.pdf) explains CO2 mixing and removal over different time scales.
- [USGS nitrate transport chapter](https://pubs.usgs.gov/circ/circ1316/html/circ1316chap4.html) describes nitrate movement through groundwater and surface water.
- [EPA inorganic metals overview](https://www.epa.gov/expobox/exposure-assessment-tools-chemical-classes-inorganics-and-fibers) describes metal persistence and condition-dependent mobility.
